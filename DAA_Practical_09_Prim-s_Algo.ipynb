```python
def prim(graph, start):
    n = len(graph)
    visited = [False] * n
    visited[start] = True
    mst = []
    cost = 0

    for _ in range(n - 1):
        minimum = float('inf')
        u = -1
        v = -1

        for i in range(n):
            if visited[i]:
                for j in range(n):
                    if not visited[j] and graph[i][j] != 0:
                        if graph[i][j] < minimum:
                            minimum = graph[i][j]
                            u = i
                            v = j

        if v == -1:
            return

        mst.append((u, v, minimum))
        cost += minimum
        visited[v] = True

    print("Edges in MST:")
    for u, v, w in mst:
        print(u, "-", v, ":", w)
    print("Minimum Cost:", cost)


graph = [
    [0, 2, 0, 6, 0],
    [2, 0, 3, 8, 5],
    [0, 3, 0, 0, 7],
    [6, 8, 0, 0, 9],
    [0, 5, 7, 9, 0]
]

prim(graph, 0)
```
